# Filter Data for Nigeria

The original dataset contains development indicators for countries and regions worldwide. Since this project focuses specifically on Nigeria, the dataset will first be filtered using Nigeria's country code, `NGA`.

This reduces the dataset to Nigeria's development indicators and allows the analysis to focus on the country's population, economic, infrastructure, and environmental conditions.

```python
nigeria_df = df[df["Country Code"] == "NGA"].copy()
```

```python
nigeria_df.shape
```

```python
nigeria_df.head()
```

# Select Relevant Indicators

The Nigeria dataset contains a wide range of development indicators covering areas such as population, economic performance, infrastructure, health, and the environment.

For this project, a focused set of indicators will be selected based on the project's objective of examining the relationship between population pressure, economic development, infrastructure access, and environmental conditions in Nigeria.

The selected indicators will be used for the subsequent data cleaning, analysis, visualization, and Power BI dashboard development.

```python
selected_indicators = [
    "Population, total",
    "Population growth (annual %)",
    "GDP per capita (current US$)",
    "GDP growth (annual %)",
    "Access to electricity (% of population)",
    "Carbon dioxide (CO2) emissions excluding LULUCF per capita (t CO2e/capita)",
    "Forest area (% of land area)",
    "Life expectancy at birth, total (years)"
]

nigeria_selected = nigeria_df[
    nigeria_df["Indicator Name"].isin(selected_indicators)
].copy()
```

# Verify Selected Indicators

The selected dataset will now be checked to confirm that all indicators required for the analysis were successfully extracted from the Nigeria dataset.

This verification step helps ensure that no indicator was excluded because of a mismatch between the selected name and the exact name used in the World Bank dataset.

```python
nigeria_selected["Indicator Name"].unique()
```

```python
nigeria_selected.shape
```

# Reshape Data for Analysis

The selected indicators are currently stored in a wide format, where each year is represented as a separate column. To make the data easier to analyze and visualize, the dataset will be reshaped into a long format.

In the long format, each row will represent one indicator for one year, with the following structure:

Country | Indicator | Year | Value

This structure is more suitable for time-series analysis, statistical calculations, and Power BI visualizations.

```python
# Reshape the dataset from wide format to long format

data = nigeria_selected.melt(
    id_vars=["Country Name", "Country Code", "Indicator Name", "Indicator Code"],
    var_name="Year",
    value_name="Value"
)

data.head()
```

# Check Data Types and Missing Values

The reshaped dataset will now be checked for data types and missing values. Since the World Bank dataset contains indicators with different data availability periods, some years may have no recorded value.

Understanding the missing data pattern is important before analysis because each indicator may cover a different range of years. We will inspect the missing values first rather than automatically deleting rows and potentially losing useful historical data.

```python
# Convert Year to numeric
data["Year"] = pd.to_numeric(data["Year"])

# Check data types
data.dtypes
```

```python
# Check missing values by indicator
data.groupby("Indicator Name")["Value"].apply(lambda x: x.isna().sum())
```

# Examine Indicator Coverage Over Time

The missing-value analysis showed that the indicators do not all have data available for the same years. This step examines the earliest and latest available year for each indicator.

Understanding the coverage period will help ensure that trends and comparisons are based on the actual years for which data is available.

```python
coverage = data.groupby("Indicator Name")["Year"].agg(
    First_Year="min",
    Last_Year="max",
    Available_Years="count"
)

coverage
```

# Check for Duplicate Records

Duplicate records can distort calculations and visualizations by counting the same indicator-year observation more than once.

We will check for duplicates using the combination of country, indicator, and year, since each indicator should have only one value for Nigeria in each year.

```python
# Check for duplicate country-indicator-year records

data.duplicated(
    subset=["Country Name", "Indicator Name", "Year"]
).sum()
```

```python
data.info()
```

# Prepare the Analysis Dataset

The dataset contains a small number of missing observations across some indicators. Since these records do not contain a numerical value for analysis, they will be excluded from calculations and visualizations.

The original dataset remains unchanged, while the cleaned analysis dataset contains only valid indicator values.

```python
# Remove rows with missing indicator values
data_clean = data.dropna(subset=["Value"]).copy()

# Check the cleaned dataset
data_clean.head()
```

```python
data_clean.shape
```

# Finalize the Clean Analysis Dataset

After removing records with missing indicator values, the original row index is no longer sequential because some rows were excluded.

We will reset the index to create a clean, continuous dataset for the analysis that follows.

```python
# Reset the index after removing missing values

data_clean = data_clean.reset_index(drop=True)

# Preview the finalized dataset
data_clean.head()
```

## 6. Comparing Development Indicators Over Time

```python
data_clean = data_clean.sort_values(
    ["Indicator Name", "Year"]
)
```

```python
# Get the earliest and latest available value for each indicator

comparison = data_clean.groupby("Indicator Name").agg(
    Earliest_Year=("Year", "min"),
    Latest_Year=("Year", "max"),
    Earliest_Value=("Value", "first"),
    Latest_Value=("Value", "last")
).reset_index()

comparison
```

## 7. Percentage Change in Development Indicators

```python
comparison["Percentage_Change"] = (
    (comparison["Latest_Value"] - comparison["Earliest_Value"])
    / comparison["Earliest_Value"]
) * 100

comparison[
    ["Indicator Name", "Earliest_Year", "Latest_Year", "Percentage_Change"]
].round(2)
```

## 8. Summary of Long-Term Development Changes

```python
summary_table = comparison[
    [
        "Indicator Name",
        "Earliest_Year",
        "Latest_Year",
        "Earliest_Value",
        "Latest_Value",
        "Percentage_Change"
    ]
].copy()

summary_table.round(2)
```

## 9. Long-Term Change in Selected Development Indicators

```python
# Exclude GDP growth because percentage change in a rate can be misleading

visual_comparison = comparison[
    comparison["Indicator Name"] != "GDP growth (annual %)"
].copy()

plt.figure(figsize=(12, 6))

plt.bar(
    visual_comparison["Indicator Name"],
    visual_comparison["Percentage_Change"]
)

plt.title("Long-Term Change in Nigeria's Development Indicators")
plt.xlabel("Development Indicator")
plt.ylabel("Percentage Change (%)")
plt.xticks(rotation=45, ha="right")
plt.axhline(0, linewidth=0.8)
plt.tight_layout()
plt.savefig("Long-Term Change in Nigeria's Development Indicators.png", dpi=300, bbox_inches="tight")

plt.show()
```

## Key Insights and Findings

Based on the analysis of Nigeria's development indicators, the following key findings were identified:

- The analysis of Nigeria's development indicators from 1960 to 2025 reveals significant long-term changes across economic, social, demographic, infrastructure, and environmental dimensions.

- Nigeria's population increased substantially from approximately 45.1 million in 1960 to 237.5 million in 2025. This represents an increase of more than five times and has contributed to growing demand for infrastructure, healthcare, education, employment, and other public services.

- GDP per capita increased from approximately $93.14 in 1960 to $1,224.25 in 2025. Although this represents substantial long-term growth, the indicator should be interpreted alongside population growth and other economic factors.

- Access to electricity increased from 27.3% of the population in 1990 to 62.5% in 2024. Despite this improvement, more than one-third of the population still lacked access to electricity in the latest available year.

- Life expectancy at birth increased from approximately 37.2 years in 1960 to 54.6 years in 2024, indicating a long-term improvement in the country's health outcomes.

- Forest area declined from approximately 29.1% of total land area in 1990 to 23.2% in 2023. This represents a reduction in forest coverage and highlights a continuing environmental sustainability concern.

- CO₂ emissions per capita increased from approximately 0.52 tonnes in 1970 to 0.57 tonnes in 2024. Although the increase was relatively small, it indicates a rise in per-capita carbon emissions over the available period.

- Nigeria's population growth remained positive over the period analyzed, reinforcing the importance of expanding infrastructure, public services, economic opportunities, and essential utilities to meet the needs of a growing population.

# Recommendations

Based on the findings from the analysis, the following recommendations are proposed:

- **Expand electricity access:** Continued investment in electricity infrastructure, particularly in underserved communities, is needed to reduce the proportion of the population without reliable access to electricity.

- **Support sustainable economic growth:** Economic policies should focus on improving productivity, employment opportunities, and living standards so that economic growth translates into meaningful improvements in income per person.

- **Strengthen environmental protection:** The decline in forest area highlights the need for stronger forest conservation, reforestation, and sustainable land-use policies.

- **Plan for population growth:** Nigeria's substantial population growth requires long-term planning for infrastructure, healthcare, education, housing, and employment to ensure that public services keep pace with demographic changes.

- **Promote low-carbon development:** As CO₂ emissions per capita have increased over the period analyzed, greater investment in renewable energy, energy efficiency, and cleaner technologies could help support economic development while limiting environmental pressure.

- **Continue investment in public health:** The long-term improvement in life expectancy should be supported through continued investment in healthcare access, disease prevention, sanitation, and other factors that contribute to population health.
